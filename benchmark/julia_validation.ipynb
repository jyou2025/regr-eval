## 11. Case 9 — Iris data

The model predicts sepal width from sepal length using all 150 Iris observations.

```julia
case_9 = data["case 9"]
display(case_9["A1:B151"])

sepal_length_9 = Float64.(vec(case_9["A2:A151"]))
observed_9 = Float64.(vec(case_9["B2:B151"]))
design_matrix_9 = hcat(ones(length(sepal_length_9)), sepal_length_9)
coefficients_9 = design_matrix_9 \ observed_9
intercept_9, slope_9 = coefficients_9
predicted_9 = design_matrix_9 * coefficients_9
display((intercept=intercept_9, slope=slope_9))

result_case_9 = evaluate(observed_9, predicted_9)
result_case_9
```
