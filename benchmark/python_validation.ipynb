## 11. Case 9 — Iris data

The model predicts sepal width from sepal length using all 150 Iris observations.

```python
case_9 = data['case 9']
display(case_9)

sepal_length_9 = case_9['sepal_length'].to_numpy()
observed_9 = case_9['sepal_width'].to_numpy()
design_matrix_9 = np.column_stack((np.ones(len(case_9)), sepal_length_9))
intercept_9, slope_9 = np.linalg.lstsq(
    design_matrix_9, observed_9, rcond=None
)[0]
predicted_9 = design_matrix_9 @ np.array([intercept_9, slope_9])
display({'intercept': intercept_9, 'slope': slope_9})

result_case_9 = rge.evaluate(observed_9, predicted_9)
result_case_9
```
